## Logging Open Hardware data

You must first install Open Hardware Monitor and then change the dll path to the dll path where the new Open Hardware Monitor is installed to do any sort of logging

Additionally, you must also install the correct version of clr from pythonnet and ensure you run vscode as adminstrator or else the code will run but log values of None type

```python
import clr
import time
import csv

# Load the LibreHardwareMonitor Library
dll_path = r"C:\Users\Satveer\Downloads\LibreHardwareMonitor-net472\LibreHardwareMonitorLib.dll"  # Update with the actual path
clr.AddReference(dll_path)

from LibreHardwareMonitor.Hardware import Computer, HardwareType, SensorType

# Initialize the computer object and enable CPU and GPU monitoring
computer = Computer()
computer.IsCpuEnabled = True
computer.IsGpuEnabled = True
computer.Open()

# Set up the CSV file for logging
with open("hardware_monitor_gpu_gpu_stress_furmark_exp3.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Timestamp", "CPU Package Power (W)", "GPU Power (W)"])

    # Log data every second for a specified duration (e.g., 30 seconds)
    duration = 330  # in seconds
    for _ in range(duration):
        cpu_power = None
        gpu_power = None

        # Update hardware data                                                                      
        for hardware in computer.Hardware:
            hardware.Update()  # Refresh sensor data

            # Get CPU Package 
            if hardware.HardwareType == HardwareType.GpuNvidia:
                for sensor in hardware.Sensors:
                    if sensor.SensorType == SensorType.Power and "GPU Package" in sensor.Name:
                        cpu_power = sensor.Value
    
                
        # Write data to the CSV file
        timestamp = time.strftime("%Y-%m-%d %H:%M:%S")
        writer.writerow([timestamp, cpu_power if cpu_power is not None else "N/A"])
        print(f"{timestamp} - GPU Package Power: {cpu_power} W")
        time.sleep(0.9)
```
