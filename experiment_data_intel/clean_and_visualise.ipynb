## Intel Power Gadget data cleaning and transforming

This function below is only used to clean and transform Intel Power Gadget data files into a overall file where all experiment result are concatenated and stored. The first 5 minutes for each experiment is concatanted and stored so that we can compare 15 minutes worth of experiment data.

```python
#Code to clean intel gadget properties and transform to overall, use this cell only if you want to extract the data into a overall csv file
import csv
import bisect

def extract_time_and_energy(file_name):
    with open(file_name, 'r') as file:
        reader = csv.reader(file)
        headers = next(reader)  

        
        time_index = headers.index("Elapsed Time (sec)")
        energy_index = headers.index("Processor Power_0(Watt)")

    
        data = [
            (float(row[time_index].strip()), float(row[energy_index].strip()))
            for row in reader
            if len(row) > max(time_index, energy_index) and row[time_index].strip() and row[energy_index].strip()
        ]
    return data

def find_closest_power(target_time, times, powers):
        pos = bisect.bisect_left(times, target_time)
        if pos == 0:
            return powers[0]
        if pos == len(times):
            return powers[-1]
        before, after = times[pos - 1], times[pos]
        return powers[pos - 1] if abs(target_time - before) < abs(target_time - after) else powers[pos]

def extract_all_intel_gadget_data(file1, file2, file3, file4):
    # Path to your CSV file

    data1 = extract_time_and_energy(file1)
    data2 = extract_time_and_energy(file2)
    data3 = extract_time_and_energy(file3)


    data1.sort(key=lambda x: x[0])
    data2.sort(key=lambda x: x[0])
    data3.sort(key=lambda x: x[0])


    times1 = [time for time, _ in data1]
    powers1 = [energy for _, energy in data1]

    times2 = [time for time, _ in data2]
    powers2 = [energy for _, energy in data2]

    times3 = [time for time, _ in data3]
    powers3 = [energy for _, energy in data3]

    # Extract Processor Power_0(Watt) every second for 300 seconds
    result1 = []
    result2 = []
    result3 = []
    for second in range(300):
        power_value1 = find_closest_power(second, times1, powers1)
        result1.append(power_value1)

        power_value2 = find_closest_power(second, times2, powers2)
        result2.append(power_value2)

        power_value3 = find_closest_power(second, times3, powers3)
        result3.append(power_value3)
    
    stacked_list = result1 + result2 + result3
    stacked_list_formatted = [[value] for value in stacked_list]

    with open(file4, mode='w', newline='') as file:
        writer = csv.writer(file)
        # Write the header
        writer.writerow(["CPU Package Power (W)"])
        # Write the stacked data
        writer.writerows(stacked_list_formatted)

    print(f"Data successfully written to {file4}")

extract_all_intel_gadget_data(r'intel_tests\memory_1000_750_rate\intel_gadget\intel_gadget_memory_1000_750_exp1.csv', r'intel_tests\memory_1000_750_rate\intel_gadget\intel_gadget_memory_1000_750_exp2.csv', r'intel_tests\memory_1000_750_rate\intel_gadget\intel_gadget_memory_1000_750_exp3.csv', r'intel_overall_experiments\intel_gadget\intel_gadget_overall_memory_1000_750_rate.CSV') 
```

```python
# you can ignore this cell
import matplotlib.pyplot as plt
import csv
import numpy as np
from scipy.stats import norm

def read_file(file_name):
    array_test = []
    with open(file_name, mode="r", newline="") as file:
        csv_reader = csv.reader(file, delimiter=' ')
        next(csv_reader, None)
        for row in csv_reader:
            array_test.append(row[0])

    array_test = [float(x) for x in array_test]

    return array_test

def create_graph(hardware_type, array, data_name):
    mu, std= norm.fit(array)
    x = np.linspace(min(array) - 10, max(array) + 10, 1000)
    gaussian_pdf = norm.pdf(x, mu, std)

    plt.figure(figsize=(10, 6))
    plt.plot(x, gaussian_pdf, label=f'Gaussian Fit\nMean={mu:.2f}, Std={std:.2f}', color='blue')
    plt.title(f"Gaussian Curve for {hardware_type} {data_name} Data")
    plt.xlabel("Power Consumption (W)")
    plt.ylabel("Density")
    plt.legend()
    plt.show()
```

The tool below is to get a multi gaussian distribution graph of a the 3 experiments done for a single test(ie, 3 experiments for laptop idle state experiments)

```python
from scipy.stats import norm

def get_gaussian_multi(file1, file2, file3, title, baseline):
    new_file = pd.read_csv(file1)
    new_file2 = pd.read_csv(file2)
    new_file3 = pd.read_csv(file3)

    mu = new_file['CPU Package Power (W)'].mean()
    std = new_file['CPU Package Power (W)'].std()

    mu_test = new_file2['CPU Package Power (W)'].mean()
    std_test = new_file2['CPU Package Power (W)'].std()

    mu_test_2 = new_file3['CPU Package Power (W)'].mean()
    std_test_2 = new_file3['CPU Package Power (W)'].std()

    print(f'Mean of hardware monitor: {mu}')
    print(f'std of hardware monitor: {std}')

    print(f'Mean of hwinfo: {mu_test}')
    print(f'std of hwinfo: {std_test}')

    print(f'Mean of {baseline}: {mu_test_2}')
    print(f'std of {baseline}: {std_test_2}')
    
    x_values_sfw = np.linspace(new_file['CPU Package Power (W)'].min(), new_file['CPU Package Power (W)'].max(), 1000)
    y_values_sfw = norm.pdf(x_values_sfw, mu, std)

    x_values = np.linspace(new_file2['CPU Package Power (W)'].min(), new_file2['CPU Package Power (W)'].max(), 1000)
    y_values = norm.pdf(x_values, mu_test, std_test)

    x_values_intel = np.linspace(new_file3['CPU Package Power (W)'].min(), new_file3['CPU Package Power (W)'].max(), 1000)
    y_values_intel = norm.pdf(x_values_intel, mu_test_2, std_test_2)

    plt.figure(figsize=(8, 5))
    plt.plot(x_values_sfw, y_values_sfw, label="Gaussian Curve hardware monitor", color="blue")
    plt.plot(x_values, y_values, label="Gaussian Curve hwinfo", color="red")
    plt.plot(x_values_intel, y_values_intel, label=f"Gaussian Curve {baseline}", color="green")
    plt.title(f"gaussian graph for {title}")
    plt.xlabel("energy consumption(W)")
    plt.ylabel("density")
    plt.grid(True)
    plt.legend()
    plt.show()
```

The tool below is to get the RMSE values for 3 files. The 3 files will essentially be overall experiment files for each test. (ie file1 will be overall experiments results for laptop idle in HWiNFO, file2 will be overall experiments results for laptop idle in OHM, etc..)

```python
import numpy as np
import pandas as pd
import csv

def get_RMSE(file1, file2, file3):
    new_graph = pd.read_csv(file1)
    np_array = new_graph.to_numpy()

    new_graph_secondary = pd.read_csv(file2)
    np_array_secondary = new_graph_secondary.to_numpy()

    new_graph_main = pd.read_csv(file3)
    np_array_main = new_graph_main.to_numpy()

    norm_factor = np.max(np_array_main) - np.min(np_array_main)

    rmse_hardware_monitor = np.sqrt(np.mean((np_array - np_array_main) ** 2))
    rmse_hwinfo = np.sqrt(np.mean((np_array_secondary - np_array_main) ** 2))
    
    return rmse_hardware_monitor/norm_factor, rmse_hwinfo/norm_factor
```
