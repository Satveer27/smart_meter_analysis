## Usability testing bar chart for HWiNFO

The code below will plot the usability testing results for HWiNFO

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV
df = pd.read_csv("usability_results.csv")
df.columns = df.columns.str.strip()  # Clean column names

# Define 'a' columns
a_cols = [f"{i}a" for i in range(1, 12)]

# Calculate average scores
a_avgs = df[a_cols].mean()

# Plot
plt.figure(figsize=(10, 6))
plt.bar(a_avgs.index, a_avgs.values, color='blue')
plt.ylim(0, 5)
plt.xlabel('Questions')
plt.ylabel('Average Likert Score')
plt.title('Average Likert Scores for Hwinfo')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
