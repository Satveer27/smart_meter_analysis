The tool below is to get a multi gaussian distribution graph of a the 3 experiments done for a single test(ie, 3 experiments for laptop idle state experiments)

```python
from scipy.stats import norm

def get_gaussian_multi(file1, file2, file3, title, baseline):
    new_file = pd.read_csv(file1)
    new_file2 = pd.read_csv(file2)
    new_file3 = pd.read_csv(file3)

    mu = new_file['CPU Package Power (W)'].mean()
    std = new_file['CPU Package Power (W)'].std()

    mu_test = new_file2['CPU Package Power (W)'].mean()
    std_test = new_file2['CPU Package Power (W)'].std()

    mu_test_2 = new_file3['CPU Package Power (W)'].mean()
    std_test_2 = new_file3['CPU Package Power (W)'].std()

    print(f'Mean of hardware monitor: {mu}')
    print(f'std of hardware monitor: {std}')

    print(f'Mean of hwinfo: {mu_test}')
    print(f'std of hwinfo: {std_test}')

    print(f'Mean of {baseline}: {mu_test_2}')
    print(f'std of {baseline}: {std_test_2}')
    
    x_values_sfw = np.linspace(new_file['CPU Package Power (W)'].min(), new_file['CPU Package Power (W)'].max(), 1000)
    y_values_sfw = norm.pdf(x_values_sfw, mu, std)

    x_values = np.linspace(new_file2['CPU Package Power (W)'].min(), new_file2['CPU Package Power (W)'].max(), 1000)
    y_values = norm.pdf(x_values, mu_test, std_test)

    x_values_intel = np.linspace(new_file3['CPU Package Power (W)'].min(), new_file3['CPU Package Power (W)'].max(), 1000)
    y_values_intel = norm.pdf(x_values_intel, mu_test_2, std_test_2)

    plt.figure(figsize=(8, 5))
    plt.plot(x_values_sfw, y_values_sfw, label="Gaussian Curve hardware monitor", color="blue")
    plt.plot(x_values, y_values, label="Gaussian Curve hwinfo", color="red")
    plt.plot(x_values_intel, y_values_intel, label=f"Gaussian Curve {baseline}", color="green")
    plt.title(f"gaussian graph for {title}")
    plt.xlabel("energy consumption(W)")
    plt.ylabel("density")
    plt.grid(True)
    plt.legend()
    plt.show()
```

The tool below is to get a multi cumulative distribution function graph of a the 3 experiments done for a single test(ie, 3 experiments for laptop idle state experiments)

```python
def get_cdf_multi(file1, file2, file3, title, baseline):
    new_file = pd.read_csv(file1)
    new_file2 = pd.read_csv(file2)
    new_file3 = pd.read_csv(file3)

    hardware_monitor_sorted = new_file["CPU Package Power (W)"].sort_values()
    hwinfo_sorted = new_file2["CPU Package Power (W)"].sort_values()
    intel_gadget_sorted = new_file3["CPU Package Power (W)"].sort_values()

    hardware_monitor_cdf = np.arange(1, len(hardware_monitor_sorted) + 1) / len(hardware_monitor_sorted)
    hwinfo_cdf = np.arange(1, len(hwinfo_sorted) + 1) / len(hwinfo_sorted)
    intel_gadget_cdf = np.arange(1, len(intel_gadget_sorted) + 1) / len(intel_gadget_sorted)

    plt.figure(figsize=(8, 5))
    plt.plot(hardware_monitor_sorted, hardware_monitor_cdf, label="CDF hardware monitor", color="blue")
    plt.plot(hwinfo_sorted, hwinfo_cdf, label="CDF hwinfo", color="red")
    plt.plot(intel_gadget_sorted, intel_gadget_cdf, label=f"CDF {baseline}", color="green")
    plt.title(f"CDF graph for {title}")
    plt.xlabel("energy consumption(W)")
    plt.ylabel("cumlative probability")
    plt.grid(True)
    plt.legend()
    plt.show()
```

The tool below is to get the RMSE values for 3 files. The 3 files will essentially be overall experiment files for each test. (ie file1 will be overall experiments results for laptop idle in HWiNFO, file2 will be overall experiments results for laptop idle in OHM, etc..)

```python
import numpy as np
import pandas as pd
import csv

def get_RMSE(file1, file2, file3):
    new_graph = pd.read_csv(file1)
    np_array = new_graph.to_numpy()

    new_graph_secondary = pd.read_csv(file2)
    np_array_secondary = new_graph_secondary.to_numpy()

    new_graph_main = pd.read_csv(file3)
    np_array_main = new_graph_main.to_numpy()

    norm_factor = np.max(np_array_main) - np.min(np_array_main)

    rmse_hardware_monitor = np.sqrt(np.mean((np_array - np_array_main) ** 2))
    rmse_hwinfo = np.sqrt(np.mean((np_array_secondary - np_array_main) ** 2))
    
    return rmse_hardware_monitor/norm_factor, rmse_hwinfo/norm_factor
```

## RMSE bar chart results for the GPU

This code below will provide the RMSE results for the GPU energy consumption

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import csv

def get_RMSE(file1, file2, file3):
    new_graph = pd.read_csv(file1)
    np_array = new_graph.to_numpy()

    new_graph_secondary = pd.read_csv(file2)
    np_array_secondary = new_graph_secondary.to_numpy()

    new_graph_main = pd.read_csv(file3)
    np_array_main = new_graph_main.to_numpy()

    norm_factor = np.max(np_array_main) - np.min(np_array_main)

    rmse_hardware_monitor = np.sqrt(np.mean((np_array - np_array_main) ** 2))
    rmse_hwinfo = np.sqrt(np.mean((np_array_secondary - np_array_main) ** 2))
    
    return rmse_hardware_monitor/norm_factor, rmse_hwinfo/norm_factor


def get_rmse_for_all_graph():
    rmse_game_hardware, rmse_game_hwinfo = get_RMSE("amd_overall_experiments_gpu/open_hardware/ohm_game_gpu_overall.csv", "amd_overall_experiments_gpu/hwinfo/hwinfo_gpu_game_overall.CSV", "amd_overall_experiments_gpu/nvidia_smi/nvidia_game_gpu_overall.csv")
    rmse_game_hardware1, rmse_game_hwinfo1 = get_RMSE("amd_overall_experiments_gpu/open_hardware/ohm_idle_gpu_overall.csv", "amd_overall_experiments_gpu/hwinfo/hwinfo_gpu_idle_overall.CSV", "amd_overall_experiments_gpu/nvidia_smi/nvidia_idle_gpu_overall.csv")
    rmse_game_hardware2, rmse_game_hwinfo2 = get_RMSE("amd_overall_experiments_gpu/open_hardware/ohm_stress_test_gpu_overall.csv", "amd_overall_experiments_gpu/hwinfo/hwinfo_gpu_stress_test_overall.CSV", "amd_overall_experiments_gpu/nvidia_smi/nvidia_stress_test_gpu_overall.csv")
    rmse_game_hardware3, rmse_game_hwinfo3 = get_RMSE("amd_overall_experiments_gpu/open_hardware/ohm_stress_furmark_gpu_overall.csv", "amd_overall_experiments_gpu/hwinfo/hwinfo_gpu_stress_furmark_overall.CSV", "amd_overall_experiments_gpu/nvidia_smi/nvidia_stress_furmark_gpu_overall.csv")
    

    # Plot bar chart
    experiments = ["idle", "game", "GPU stressor with heavy load", "GPU stressor with furmark"]
    rmse_values = [rmse_game_hardware, rmse_game_hardware1, rmse_game_hardware2, rmse_game_hardware3]
    
    rmse_values_2 = [rmse_game_hwinfo, rmse_game_hwinfo1, rmse_game_hwinfo2, rmse_game_hwinfo3]
    
    x = np.arange(len(experiments))
    width = 0.35

    fig, ax = plt.subplots(figsize=(12, 6))
    ax.bar(x - width/2, rmse_values, width, label='Hardware Monitor', color='royalblue')
    ax.bar(x + width/2, rmse_values_2, width, label='HWInfo', color='orange')

    ax.set_xlabel("Experiments")
    ax.set_ylabel("normalized RMSE")
    ax.set_title("normalized RMSE Comparison Across Experiments")
    ax.set_xticks(x)
    ax.set_xticklabels(experiments, rotation=45, ha="right")
    ax.legend()
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Show plot
    plt.show()

get_rmse_for_all_graph()
```
